## Part 1: Meet the Members of the Royal Court

Before building the royal court, let's understand how inheritance works.

A *Prince* and a *Knight* may have different responsibilities, but both belong to the kingdom. They might both have a name, age and kingdom.

Instead of writing these attributes separately in both classes, we can create a common parent class called `RoyalMember`.

````python
class RoyalMember:
    def __init__(self, name, age, kingdom):
        self.name = name
        self.age = age
        self.kingdom = kingdom

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}.")

````

````python
arthur = RoyalMember("Arthur",25,"Eldoria")
print(arthur.name)
print(arthur.age)
print(arthur.kingdom)
arthur.introduce()
````

## Part 2: The Knights Join the Kingdom (Inheritance)

The King now wants to recruit knights to protect the kingdom.
Every knight has a name, age and kingdom, just like other royal members. However, knights also have additional attributes, such as their weapon and number of battles won.

Instead of rewriting all the attributes, we can make the `Knight` class inherit from `RoyalMember`.

````python
class Knight(RoyalMember):
    def __init__(self, name, age, kingdom, weapon, battles_won):
        super().__init__(name, age, kingdom)
        self.weapon = weapon
        self.battles_won = battles_won

    def display_details(self):
        print(f"Knight: {self.name}")
        print(f"Kingdom: {self.kingdom}")
        print(f"Weapon: {self.weapon}")
        print(f"Battles won: {self.battles_won}")
````

Notice a few things:

`Knight(RoyalMember)` means Knight inherits from `RoyalMember`.

`super().__init__()` calls the constructor of the parent class.

The Knight class receives all the attributes of RoyalMember and adds two of its own.

Q: Did we need to define the introduce() method again inside the Knight class? Why or why not?

**A**:No. Knight inherits from RoyalMember, so it automatically gets all of the parent’s methods, including introduce(). When we call lancelot.introduce(), Python doesn’t find introduce() in the Knight class, so it looks in the parent class RoyalMember and uses that version. The name, age and kingdom attributes are also set by the parent’s __init__ (through super().__init__()), so introduce() works on a knight exactly as it does on any royal member. This avoids rewriting the same code, which is the main benefit of inheritance. We would only define introduce() again in Knight if we wanted knights to introduce themselves differently, which is called overriding the meth

## Part 3: A Prince Has His Own Way of Doing Things (Method Overriding)

The King has now introduced his son, Prince Edward, to the royal court.

Like knights, princes are also members of the kingdom. However, a prince introduces himself differently.

A knight might introduce himself by mentioning his title and kingdom, while a prince might mention his royal status.

We can achieve this using **method overriding**.

Method overriding occurs when a child class provides its own implementation of a method that already exists in its parent class.

````python
class Prince(RoyalMember):
    def __init__(self, name, age, kingdom, title):
        super().__init__(name, age, kingdom)
        self.title = title

    def introduce(self):
        print(f"I am {self.title} {self.name}, heir to the throne of {self.kingdom}.")
````

Notice that `introduce()` already exists in `RoyalMember`. However, the `Prince` class provides its own version.

### Task 3: Give the knights their own introductions

Currently, knights use the `introduce()` method inherited from `RoyalMember`.

Modify the `Knight` class to override this method.

Make the knight introduce himself using his name and weapon.

Mention how many battles he has won.

Create a new knight named Sir Gareth, aged 32, from Northvale, who carries an axe and has won 15 battles.

Call his `introduce()` method.

Expected introduction:

`I am Sir Gareth of Northvale. I fight with an axe and have won 15 battles.`


````python
class Knight(RoyalMember):
    def __init__(self, name, age, kingdom, weapon, battles_won):
        super().__init__(name, age, kingdom)
        self.weapon = weapon
        self.battles_won = battles_won

    def display_details(self):
        print(f"Knight: {self.name}")
        print(f"Kingdom: {self.kingdom}")
        print(f"Weapon: {self.weapon}")
        print(f"Battles won: {self.battles_won}")

    def introduce(self):   # overrides RoyalMember's version
        weapon = self.weapon.lower()
        article = "an" if weapon[0] in "aeiou" else "a"
        print(f"I am {self.name} of {self.kingdom}. "
              f"I fight with {article} {weapon} and have won {self.battles_won} battles.")
gareth = Knight("Sir Gareth", 32, "Northvale", "Axe", 15)
gareth.introduce()
````

Q: What happens when a `Knight` object calls `introduce()`? What happens when a `Prince` object calls `introduce()`? Both classes inherit from `RoyalMember`. Why do they produce different outputs?

A: When a Knight object calls introduce(), Python first looks in the Knight class. Since Knight now defines its own introduce(), that version runs, and the knight talks about his weapon and battles. When a Prince object calls introduce(), Python looks in the Prince class. If Prince doesn’t define its own version, Python moves up to the parent class RoyalMember and runs that one, so the prince gives the basic name, kingdom and age introduction. They produce different outputs because Knight overrides the inherited method. A child class’s own version of a method takes priority over the parent’s. This lets each subclass customize behavior while still sharing the same method name

## Part 4: One Royal Court, Many Personalities (Polymorphism)

The King has finally assembled his royal court.

It includes princes and knights, all of whom have their own versions of the `introduce()` method.

Rather than calling each object's method separately, we can store all the members in one list and use a loop.

This is an example of **polymorphism**.

Polymorphism allows different objects to respond to the same method call in their own way.

````python
members = [
    RoyalMember("William", 45, "Eldoria"),
    Knight("Lancelot", 30, "Eldoria", "Sword", 12),
    Prince("Edward", 21, "Eldoria", "Prince"),
    Knight("Cedric", 27, "Northvale", "Spear", 8)
]

for member in members:
    member.introduce()

# Even though every object receives the same method call, member.introduce(),
# the output depends on the type of object.
````

## Part 5: The Royal Tournament (Putting It All Together)

The King has announced a grand tournament! Knights from different kingdoms will compete to demonstrate their skills.

Each knight has a different number of battles won, and the King wants to identify the most experienced knight.

You will now extend your existing classes to build a small tournament management system.

### Task 5: Build the tournament system

1) Step 1: Add a method to the `Knight` class

- Create a method called `fight()` that prints:

  `Sir [name] enters the tournament with a [weapon]!`

  For example:

  `Sir Lancelot enters the tournament with a Sword!`

2) Step 2: Create the tournament participants

- Create at least four knights with different:

  - a) *Names*
  - b) *Kingdoms*
  - c) *Weapons*
  - d) *Numbers of battles won*

  Store all the knights in a list called `tournament`.

3) Step 3: Display the participants

- Use a loop to:

  - Call the `introduce()` method for every knight.
  - Call the `fight()` method for every knight.
  - Print their number of battles won.

4) Step 4: Find the most experienced knight

- Find the knight with the highest number of battles won.
- Print their name and the number of battles they have won.


````python
# STEP 1 - ADD A fight() METHOD TO THE EXISTING KNIGHT CLASS
class Knight(RoyalMember):
    def __init__(self, name, age, kingdom, weapon, battles_won):
        super().__init__(name, age, kingdom)
        self.weapon = weapon
        self.battles_won = battles_won

    def display_details(self):
        print(f"Knight: {self.name}")
        print(f"Kingdom: {self.kingdom}")
        print(f"Weapon: {self.weapon}")
        print(f"Battles won: {self.battles_won}")

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}. "
              f"My weapon is a {self.weapon} and I have won {self.battles_won} battles.")

    def fight(self):
        print(f"Sir {self.name} enters the tournament with a {self.weapon}!")


# STEP 2 - CREATE A LIST OF PARTICIPANTS CALLED tournament
tournament = [
    Knight("Lancelot", 30, "Eldoria", "Sword", 12),
    Knight("Cedric", 27, "Northvale", "Spear", 8),
    Knight("Gareth", 32, "Northvale", "Axe", 15),
    Knight("Tristan", 29, "Westmarch", "Bow", 10),
]


# STEP 3 - LOOP OVER THE LIST, CALL introduce(), fight() AND PRINT THE NUMBER OF BATTLES WON
for knight in tournament:
    knight.introduce()
    knight.fight()
    print(f"Battles won: {knight.battles_won}")
    print()   # blank line between knights


# STEP 4 - FIND THE MOST EXPERIENCED KNIGHT
champion = tournament[0]
for knight in tournament:
    if knight.battles_won > champion.battles_won:
        champion = knight

print(f"The most experienced knight is Sir {champion.name} "
      f"with {champion.battles_won} battles won!")

````

## Part 6: The King's Grand Council – The Ultimate Royal Challenge! (optional!!!)

The King has summoned all members of the kingdom for an important council meeting. Each member has different responsibilities and recommendations for improving the kingdom.

The King wants to build a **Royal Council Management System** to keep track of his council members and their advice.

### Your Tasks:

**Step 1: Create a new method**

Add a method called `give_advice()` to the `RoyalMember` class.

The method should print:

`[Name] has no official recommendation for the kingdom.`

**Step 2: Override the method**

Each type of royal member should have its own version of `give_advice()`:

- **Prince:** Recommends expanding the kingdom's territory.
- **Knight:** Recommends strengthening the kingdom's army.
- **RoyalGuard:** Recommends improving the security of the castle.

**Step 3: Assemble the council**

1. Create a list called `royal_council`.
2. Add at least one `RoyalMember`, two `Knights`, two `Princes` and two `RoyalGuards`.
3. Use a loop to make every member introduce themselves and give their advice.

**Step 4: Count the council members**

Using `isinstance()`, count how many members belong to each class.

Print the total number of:
- RoyalMembers
- Knights
- Princes
- RoyalGuards

**Step 5: The King's final decision**

The King wants to know how many council members recommended improving the kingdom's military or security.

Create a variable called `security_recommendations` and count the number of Knights and RoyalGuards in the council.

Print the final count.


````python

# Step 1: Add give_advice() to RoyalMember
class RoyalMember:
    def __init__(self, name, age, kingdom):
        self.name = name
        self.age = age
        self.kingdom = kingdom

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}.")
        print(f"I am {self.age} years old.")

    def give_advice(self):
        print(f"{self.name} has no official recommendation for the kingdom.")


# Step 2: Override the give_advice() method
class Prince(RoyalMember):
    def __init__(self, name, age, kingdom, title):
        super().__init__(name, age, kingdom)
        self.title = title

    def introduce(self):
        print(f"I am {self.title} {self.name}, heir to the throne of {self.kingdom}.")

    def give_advice(self):
        print(f"{self.title} {self.name} recommends expanding the kingdom's territory.")


class Knight(RoyalMember):
    def __init__(self, name, age, kingdom, weapon, battles_won):
        super().__init__(name, age, kingdom)
        self.weapon = weapon
        self.battles_won = battles_won

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}. "
              f"My weapon is a {self.weapon} and I have won {self.battles_won} battles.")

    def fight(self):
        print(f"Sir {self.name} enters the tournament with a {self.weapon}!")

    def give_advice(self):
        print(f"Sir {self.name} recommends strengthening the kingdom's army.")


class RoyalGuard(RoyalMember):
    def __init__(self, name, age, kingdom, post):
        super().__init__(name, age, kingdom)
        self.post = post

    def guard_duty(self):
        print(f"{self.name} is on duty at the {self.post}.")

    def introduce(self):
        print(f"I am {self.name}, a Royal Guard of {self.kingdom}. "
              f"I stand watch at the {self.post}.")

    def give_advice(self):
        print(f"{self.name} recommends improving the security of the castle.")


# Step 3: Assemble the council members
royal_council = [
    RoyalMember("William", 45, "Eldoria"),
    Knight("Lancelot", 30, "Eldoria", "Sword", 12),
    Knight("Gareth", 32, "Northvale", "Axe", 15),
    Prince("Edward", 21, "Eldoria", "Prince"),
    Prince("Henry", 18, "Northvale", "Prince"),
    RoyalGuard("Roland", 35, "Eldoria", "Castle Gate"),
    RoyalGuard("Bran", 28, "Eldoria", "Throne Room"),
]

# Introduce every member and display their advice
for member in royal_council:
    member.introduce()
    member.give_advice()
    print()


# Step 4: Count the council members
knights = princes = guards = royal_members = 0

for member in royal_council:
    if isinstance(member, Knight):
        knights += 1
    elif isinstance(member, Prince):
        princes += 1
    elif isinstance(member, RoyalGuard):
        guards += 1
    elif isinstance(member, RoyalMember):
        royal_members += 1

print(f"RoyalMembers: {royal_members}")
print(f"Knights: {knights}")
print(f"Princes: {princes}")
print(f"RoyalGuards: {guards}")


# Step 5: The King's final decision
security_recommendations = 0
for member in royal_council:
    if isinstance(member, (Knight, RoyalGuard)):
        security_recommendations += 1

print(f"\nCouncil members recommending military or security improvements: "
      f"{security_recommendations}")
````

### Final Questions:

1. Why can all the objects in `royal_council` use the same `give_advice()` method?

    All the objects can use `give_advice()` because every class in the council inherits from `RoyalMember`, and `RoyalMember` defines `give_advice()`. Since `Knight`, `Prince` and `RoyalGuard` are subclasses of `RoyalMember`, each of their objects is also a `RoyalMember` and is guaranteed to have the method, either its own version or the inherited one. That's why one loop can safely call `member.give_advice()` on every object without checking its type.
2. Which OOP concept allows each class to provide its own version of `give_advice()`?
  
  
    **Polymorphism**, achieved through **method overriding**. Each subclass defines its own `give_advice()` with the same name as the parent's, and that version replaces the parent's for objects of that class. When the method is called, Python checks the object's actual class and runs that class's version. So the same call gives different advice from a prince, a knight and a guard.

3. What would happen if the `Knight` class did not override `give_advice()`?
   
  
    Knights would fall back on the inherited version from `RoyalMember`. When Python couldn't find `give_advice()` in `Knight`, it would look in the parent class and use that method instead. Every knight would then print something like "Lancelot has no official recommendation for the kingdom." The program would still run without errors, but knights would lose their specific advice about strengthening the army.

4. Why is using inheritance helpful when building a system with several related classes?


    Inheritance lets related classes share common code instead of repeating it. Attributes like `name`, `age` and `kingdom`, and default behavior like `introduce()` and `give_advice()`, are written once in `RoyalMember`, and each subclass only adds or changes what's unique to it. This means less duplicated code, fewer places for bugs, and easier maintenance, because a fix in the parent class automatically applies to all subclasses. It also makes the system easy to extend: a new class like `RoyalGuard` can be added with just a few lines and immediately works with existing code, such as the loops over `royal_council`.
